```python
# Cell 5: Jalankan FastAPI + ngrok Static Tunnel
# URL backend PERMANEN: https://aged-breeches-splashing.ngrok-free.dev

import subprocess, threading, time, os
from pyngrok import ngrok

os.chdir('/kaggle/working/ImgTools/apps/ai')

def run_server():
    subprocess.run([
        'uvicorn', 'main:app',
        '--host', '0.0.0.0',
        '--port', '7860',
        '--workers', '1'
    ])

threading.Thread(target=run_server, daemon=True).start()
print('Server Uvicorn dimulai...')
time.sleep(5)  # Tunggu server siap

# Buka tunnel dengan static domain
tunnel = ngrok.connect(
    addr=7860,
    domain=NGROK_STATIC_DOMAIN
)

url = f"https://{NGROK_STATIC_DOMAIN}"
print()
print('=' * 65)
print('BACKEND AI AKTIF! (URL PERMANEN)')
print(f'URL Publik    : {url}')
print(f'Health Check  : {url}/health')
print(f'Swagger Docs  : {url}/docs')
print()
print('apps/web/.env.local sudah dikonfigurasi dengan URL ini!')
print(f'NEXT_PUBLIC_AI_URL={url}')
print('URL ini TIDAK berubah setiap restart!')
print('=' * 65)
```

```python
# Cell 6: Keep-Alive â€” Cegah Kaggle auto-shutdown saat idle (ping tiap 20 menit)
# Jalankan cell ini agar sesi Kaggle tidak mati otomatis
# Tekan tombol STOP untuk mematikan server

import time, urllib.request, json

print('Keep-alive aktif. Tekan Stop untuk mematikan.')
i = 0
while True:
    try:
        with urllib.request.urlopen('http://localhost:7860/health', timeout=10) as r:
            data = json.loads(r.read())
            i += 1
            print(f'[{time.strftime("%H:%M:%S")}] Ping #{i}: {data}')
    except Exception as e:
        print(f'[{time.strftime("%H:%M:%S")}] Peringatan: {e}')
    time.sleep(1200)
```
